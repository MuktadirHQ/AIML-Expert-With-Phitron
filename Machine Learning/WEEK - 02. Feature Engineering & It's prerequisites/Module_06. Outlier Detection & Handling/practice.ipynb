```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

### Task-1

```python
df = np.array(
    [18, 19, 20, 20, 21, 21, 22, 22, 23, 24, 25, 26, 27, 28, 30, 32, 35, 40, 45, 70]
)
mean_age = df.mean()
std_age = df.std()
z_score = (df - mean_age) / std_age
print(z_score)

outliers = df[np.abs(z_score) > 3]
print(outliers)

result = df[np.abs(z_score)<=3]
print(result)
```

### Task-2

```python
df = np.array(
    [25, 28, 30, 32, 32, 33, 35, 35, 36, 38, 40, 42, 45, 48, 50, 55, 60, 65, 70, 80,
120, 150, 200, 250, 500]

)

q1 = np.percentile(df, 25)
print(q1)
q3 = np.percentile(df, 75)
print(q3)

iqr = q3 - q1
print(iqr)

lower_bound = q1 - 1.5 * iqr
print(lower_bound)
upper_bound = q3 + 1.5 * iqr
print(upper_bound)

outliers = df[(df < lower_bound) | (df > upper_bound)]
print(outliers)
clean = df.clip(lower_bound, upper_bound)
print(clean)
```

```python
df = np.array(
    [12, 15, 18, 20, 22, 22, 23, 24, 25, 25, 26, 27, 28, 30, 32, 35, 38, 40, 45, 50, 55, 60, 65, 70, 80, 90, 100, 120, 150, 200, 500, 1000, 2000]
)

plt.hist(df, bins=30)
plt.show()
```

```python
mean = df.mean()
std = df.std()

z_score = (df - mean) /std
outliers = df[z_score>3]
print(outliers)

q1 = np.percentile(df, 25)
q3 = np.percentile(df, 75)
iqr = q3 - q1
lb = q1-1.5*iqr
ub = q3+1.5*iqr

outliers = df[(df<lb) | (df>ub)]
print(outliers)
```

### Task-6

```python
df = np.array(
    [5, 5, 6, 6, 7, 7, 7, 8, 8, 8, 9, 9, 10, 10, 10, 11, 12, 13, 15, 18, 20, 25, 30, 35, 40, 50, 60, 80, 100, 120, 150, 200, 300]
)

mean = df.mean()
median = np.median(df)
mode = stats.mode(df)
print(mean, median, mode)

q1 = np.percentile(df, 25)
q3 = np.percentile(df, 75)
iqr = q3 - q1
lb = q1-1.5*iqr
ub = q3+1.5*iqr

outliers = df[(df<lb) | (df>ub)]
print(outliers)

lb = np.percentile(df, 3)
ub = np.percentile(df, 97)
df = df.clip(lb, ub)
print(df)
print(min(df))
print(max(df))
print(df.mean())
```
