```python
import pandas as pd
import numpy as np
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, LabelEncoder
```

# 1. Diabetes Dataset – Missing Value Imputation & Encoding

```python
df = pd.read_csv("/content/diabetes.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
col = ['Glucose', 'BloodPressure', 'BMI']
df[col] = df[col].replace(0, np.nan)
df
```

```python
imputer = SimpleImputer(missing_values=np.nan, strategy='median')
imputer.fit(df[col])
df[col] = imputer.transform(df[col])
df
```

```python
bins = [0, 29, 50, 150]
labels = ['Young', 'Adult', 'Senior']
df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)
df
```

```python
encoder = OrdinalEncoder(categories=[['Young', 'Adult', 'Senior']])
encoder.fit(df[['Age_Group']])
df['Age_Group_Encoded'] = encoder.transform(df[['Age_Group']])
df.shape
```

# 3. Tips Dataset – Label Encoding

```python
df = sns.load_dataset("tips")
df.head()
```

```python
sex_lbl = LabelEncoder()
sex_lbl.fit(df[['sex']])
df['Encoded_sex'] = sex_lbl.transform(df[['sex']])
df.head()
```

```python
smoke_lbl = LabelEncoder()
smoke_lbl.fit(df[['smoker']])
df['Encoded_smoker'] = smoke_lbl.transform(df[['smoker']])
df.head()
```

```python
tips_bin = [0, 2.99, 6.0, float('inf')]
tips_category = ['Low', 'Medium', 'High']
df['tip_category'] = pd.cut(df['tip'], bins=tips_bin, labels=tips_category)
df
```

```python
tip_encoded = OrdinalEncoder(categories=[tips_category])
tip_encoded.fit(df[['tip_category']])
df['tip_category_encoded'] = tip_encoded.transform(df[['tip_category']])
df.sample(10)
```

# 4. Car Price Dataset – Train/Test Encoding Pipeline


```python
df = pd.read_csv("/content/CarPrice_Assignment.csv")
df
```

```python
x = df.drop(['price', 'car_ID', 'CarName'], axis=1)
y = df['price']

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
encoded_df = OneHotEncoder(categories=[['gas', 'diesel']], sparse_output=False, dtype=int)
encoded_df.fit(X_train[['fueltype']])

X_train_array = encoded_df.transform(X_train[['fueltype']])
X_test_array = encoded_df.transform(X_test[['fueltype']])

temp_X_train = pd.DataFrame(X_train_array, columns=encoded_df.get_feature_names_out(['fueltype']), index=X_train.index)
temp_X_test = pd.DataFrame(X_test_array, columns=encoded_df.get_feature_names_out(['fueltype']), index=X_test.index)

X_train = pd.concat([X_train.drop('fueltype', axis=1), temp_X_train], axis=1)
X_test = pd.concat([X_test.drop('fueltype', axis=1), temp_X_test], axis=1)
```

```python
ohe_drive = OneHotEncoder(sparse_output=False, drop='first', dtype='int')
ohe_drive.fit(X_train[['drivewheel']])

X_train_array = ohe_drive.transform(X_train[['drivewheel']])
X_test_array = ohe_drive.transform(X_test[['drivewheel']])

X_train_temp = pd.DataFrame(X_train_array, columns=ohe_drive.get_feature_names_out(['drivewheel']), index=X_train.index)
X_test_temp = pd.DataFrame(X_test_array, columns=ohe_drive.get_feature_names_out(['drivewheel']), index=X_test.index)


X_train = pd.concat([X_train.drop(['drivewheel'], axis=1), X_train_temp], axis=1)
X_test = pd.concat(
    [ X_test.drop(['drivewheel'], axis=1), X_test ],
    axis=1
)

X_test
```
