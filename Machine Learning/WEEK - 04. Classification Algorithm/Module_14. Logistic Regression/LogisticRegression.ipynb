```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def sigmoid(z):
    return 1 / (1+np.exp(-z))
```

```python
z = np.arange(-10, 11, 0.1)
sigmoid_values = sigmoid(z)
plt.plot(z, sigmoid_values)

plt.xlabel("Z")
plt.ylabel("sigmoid(z)")
plt.show()
```
