**১. Artificial Intelligence এবং Machine Learning এর মধ্যে পার্থক্য কী?**
* **Artificial Intelligence (AI):** এটি হলো একটি বিশাল ক্ষেত্র, যার মূল উদ্দেশ্য হলো মেশিন বা কম্পিউটারকে মানুষের মতো বুদ্ধিমত্তা দেওয়া, যাতে তারা মানুষের মতো সিদ্ধান্ত নিতে পারে।
* **Machine Learning (ML):** এটি AI এর একটি সাবসেট বা শাখা। এখানে স্পেসিফিক কোনো প্রোগ্রামিং না করেই বিপুল পরিমাণ ডেটা ব্যবহার করে মেশিনকে নিজে নিজে শিখতে ও সিদ্ধান্ত নিতে পারদর্শী করা হয়।

**২. ANI এবং AGI এর মধ্যে পার্থক্য লিখো।**
* **ANI (Artificial Narrow Intelligence):** এটি নির্দিষ্ট একটি কাজের জন্য তৈরি করা হয়। এটি শুধু ওই একটি কাজই ভালোভাবে করতে পারে (যেমন- দাবা খেলার বট বা স্প্যাম ফিল্টার)। একে Weak AI-ও বলা হয়।
* **AGI (Artificial General Intelligence):** এটি হলো মানুষের মতো বুদ্ধিমত্তাসম্পন্ন AI, যা যেকোনো নতুন এবং জটিল কাজ মানুষের মতোই শিখতে ও সমাধান করতে পারবে। (এটি এখনো গবেষণার পর্যায়ে আছে)।

**৩. Deep Learning কেন Machine Learning এর subset?**

Deep Learning হলো Machine Learning এর একটি বিশেষ পদ্ধতি। Machine learning-এ সাধারণত ম্যানুয়ালি ফিচার ইঞ্জিনিয়ারিং করতে হয়, কিন্তু Deep Learning মানুষের মস্তিষ্কের নিউরাল নেটওয়ার্কের (Artificial Neural Networks) অনুকরণে কাজ করে এবং বিপুল পরিমাণ ডেটা থেকে নিজে নিজেই জটিল ফিচার বা প্যাটার্ন শিখতে পারে। যেহেতু এটি ML-এরই একটি শাখা, তাই এটি ML-এর subset।

**৪. Alan Turing কেন AI history তে গুরুত্বপূর্ণ?**

অ্যালান টুরিংকে আধুনিক কম্পিউটিং এবং আর্টিফিশিয়াল ইন্টেলিজেন্সের অন্যতম জনক বলা হয়। ১৯৫০ সালে তিনি "Turing Test" আবিষ্কার করেন, যা কোনো মেশিনের মানুষের মতো বুদ্ধি বা চিন্তাশক্তি আছে কিনা, তা পরিমাপ করার একটি ঐতিহাসিক মাপকাঠি হিসেবে বিবেচিত হয়।

**৫. Real-life AI applications এর ৫টি উদাহরণ দাও।**
* ভয়েস অ্যাসিস্ট্যান্ট (যেমন: Apple Siri, Amazon Alexa)।
* রেকমেন্ডেশন সিস্টেম (যেমন: Netflix, YouTube বা E-commerce সাইটের সাজেশন)।
* স্বয়ংক্রিয় বা সেলফ-ড্রাইভিং গাড়ি (যেমন: Tesla)।
* ফেস রিকগনিশন (যেমন: মোবাইলের ফেস আনলক)।

## Types of Machine Learning
1. Supervised Learning কী?
2. Regression এবং Classification এর মধ্যে পার্থক্য লিখো।
3. House Price Prediction কোন ধরনের problem? কেন?
4. Spam Email Detection কোন ধরনের ML problem?
5. Unsupervised Learning এ labels কেন থাকে না?


**১. Supervised Learning কী?**

যেখানে মডেলকে ট্রেনিং দেওয়ার সময় ডেটার পাশাপাশি তার সঠিক উত্তর বা লেবেল (Labels) দিয়ে দেওয়া হয়, তাকে Supervised Learning বলে। মডেল এই ইনপুট ও আউটপুটের মধ্যকার সম্পর্ক শেখে।

**২. Regression এবং Classification এর মধ্যে পার্থক্য লিখো।**

* **Regression:** এটি কন্টিনিউয়াস (Continuous) বা সংখ্যাভিত্তিক ভ্যালু প্রেডিক্ট করে। (যেমন: আগামীকালের তাপমাত্রা কত হবে বা একটি বাড়ির দাম কত)।
* **Classification:** এটি ডেটাকে বিভিন্ন ক্যাটাগরিতে (Discrete) ভাগ করে। (যেমন: ইমেইলটি কি স্প্যাম নাকি স্প্যাম নয়)।

**৩. House Price Prediction কোন ধরনের problem? কেন?**

এটি একটি **Regression** প্রবলেম। কারণ বাড়ির দাম একটি সংখ্যা (যেমন- ৫০ লক্ষ বা ২ কোটি টাকা) এবং এটি কন্টিনিউয়াস ভ্যালু। এখানে আমরা কোনো নির্দিষ্ট ক্যাটাগরি বা ক্লাসে ডেটাকে ভাগ করছি না।

**৪. Spam Email Detection কোন ধরনের ML problem?**

এটি একটি **Classification** প্রবলেম (Supervised Learning এর অংশ)। কারণ এখানে আউটপুট মাত্র দুটি ক্যাটাগরিতে বিভক্ত: "Spam" অথবা "Not Spam"।

**৫. Unsupervised Learning এ labels কেন থাকে না?**

Unsupervised learning এর মূল উদ্দেশ্য হলো অবিন্যস্ত বা আনলেবেলড (unlabeled) ডেটার ভেতর থেকে লুকানো প্যাটার্ন, গঠন বা গ্রুপ (Clustering) খুঁজে বের করা। যেহেতু আমরা আগে থেকে জানি না ডেটার ভেতরে কী গ্রুপ লুকিয়ে আছে, তাই এখানে কোনো লেবেল বা সঠিক উত্তর দেওয়া থাকে না।

## ML Pipeline
1. Feature Engineering কী?
2. Train-Test Split কেন প্রয়োজন?
3. Data Preprocessing এর ৩টি কাজ লিখো।
4. Hyperparameter এবং Parameter এর মধ্যে পার্থক্য কী?
5. Model Drift বলতে কী বুঝায়?


**১. Feature Engineering কী?**

র (Raw) ডেটাসেট থেকে নতুন দরকারি বৈশিষ্ট্য (features) তৈরি করা বা অপ্রয়োজনীয় বৈশিষ্ট্য বাদ দিয়ে মডেলের শেখার উপযোগী ডেটা তৈরি করার প্রক্রিয়াকেই Feature Engineering বলে। এটি মডেলের একুরেসি বাড়াতে অনেক সাহায্য করে।

**২. Train-Test Split কেন প্রয়োজন?**

মডেলটি শুধু মুখস্থ করছে নাকি সত্যিই শিখছে, তা যাচাই করার জন্য ডেটাসেটকে দুই ভাগে ভাগ করা হয়। Train ডেটা দিয়ে মডেলকে শেখানো হয় এবং Test বা অদেখা ডেটা দিয়ে পরীক্ষা করা হয় যে মডেলটি নতুন পরিস্থিতিতে কতটুকু ভালো কাজ করতে পারছে।

**৩. Data Preprocessing এর ৩টি কাজ লিখো।**

* Missing values (খালি জায়গা) পূরণ করা বা বাদ দেওয়া।
* ক্যাটাগরিকাল ডেটাকে (যেমন- Male/Female) নাম্বারে (0/1) এনকোডিং করা।
* ডেটা স্কেলিং বা নরমালাইজেশন করা (সব ডেটাকে একটি নির্দিষ্ট রেঞ্জের মধ্যে নিয়ে আসা)।

**৪. Hyperparameter এবং Parameter এর মধ্যে পার্থক্য কী?**

* **Parameter:** মডেল ট্রেনিংয়ের সময় ডেটা থেকে নিজে নিজে যে মানগুলো শেখে (যেমন- Linear Regression এর m এবং b)।
* **Hyperparameter:** মডেল ট্রেনিং শুরু করার আগে ডেভেলপার বা ইঞ্জিনিয়ারকে ম্যানুয়ালি যে মানগুলো সেট করে দিতে হয় (যেমন- Learning rate, Number of epochs)।

**৫. Model Drift বলতে কী বুঝায়?**

সময়ের সাথে সাথে বাস্তব দুনিয়ার ডেটার প্যাটার্ন পরিবর্তন হওয়ার কারণে যখন আগে থেকে ট্রেইন করা কোনো মডেলের প্রেডিকশন ক্ষমতা বা পারফরম্যান্স ধীরে ধীরে কমতে থাকে, তখন তাকে Model Drift (বা Data Drift) বলে।

## Simple Linear Regression
1. Simple Linear Regression কী predict করে?
2. Independent Variable এবং Dependent Variable এর পার্থক্য লিখো।
3. y = mx + b equation এ m এবং b কী বোঝায়?
4. Best Fit Line বলতে কী বুঝায়?
5. Linear Regression এর একটি real-life use case লিখো।


**১. Simple Linear Regression কী predict করে?**

এটি একটিমাত্র স্বাধীন (Independent) ভেরিয়েবলের ওপর ভিত্তি করে একটি নির্ভরশীল (Dependent) ভেরিয়েবলের সংখ্যাভিত্তিক বা কন্টিনিউয়াস ভ্যালু প্রেডিক্ট করে।

**২. Independent Variable এবং Dependent Variable এর পার্থক্য লিখো।**

* **Independent Variable (x):** এটি ইনপুট ভেরিয়েবল, যার মান আমাদের জানা থাকে বা আমরা মডেলকে দিই। এটি অন্য কোনো কিছুর ওপর নির্ভর করে না।
* **Dependent Variable (y):** এটি আউটপুট ভেরিয়েবল, যার মান আমরা প্রেডিক্ট করতে বা বের করতে চাই। এর মান x এর ওপর নির্ভর করে।

**৩. y = mx + b equation এ m এবং b কী বোঝায়?**

* **m (Slope/ঢাল):** x এর মান ১ ইউনিট পরিবর্তন হলে y এর মান কতটুকু পরিবর্তন হবে তা নির্দেশ করে।
* **b (Y-intercept):** যখন x = 0, তখন y এর মান কত হবে, এটি তা নির্দেশ করে। এটি গ্রাফের Y-অক্ষকে ছেদ করার বিন্দু।

**৪. Best Fit Line বলতে কী বুঝায়?**

এটি এমন একটি সরলরেখা যা ডেটাসেটের পয়েন্টগুলোর মাঝখান দিয়ে এমনভাবে যায় যেন, প্রতিটি ডেটা পয়েন্ট থেকে ওই রেখার লম্ব দূরত্ব (Error) সব মিলিয়ে সবচেয়ে কম হয়। এটিই আমাদের মডেলের তৈরি করা রেখা।

**৫. Linear Regression এর একটি real-life use case লিখো।**

কর্মীর অভিজ্ঞতা (Years of Experience) জানা থাকলে তার বেতন (Salary) প্রেডিক্ট করা।



## Cost Function
1. Cost Function কেন দরকার?
2. Prediction Error কী?
3. Model error কমানো কেন গুরুত্বপূর্ণ?
4. Large error হলে Cost Function এর value কেমন হবে?
5. Cost কমানোর process এর নাম কী?


**১. Cost Function কেন দরকার?**

মডেলটি তার প্রেডিকশনে কতটা ভুল করছে তা গাণিতিকভাবে পরিমাপ করার জন্য Cost Function (যেমন- Mean Squared Error) ব্যবহার করা হয়।

**২. Prediction Error কী?**

ডেটাসেটে থাকা আসল ভ্যালু (Actual value) এবং মডেলের প্রেডিক্ট করা ভ্যালুর (Predicted value) মাঝখানের পার্থক্যই হলো Prediction Error।

**৩. Model error কমানো কেন গুরুত্বপূর্ণ?**

মডেলের Error যত কমবে, মডেলের Best fit line তত নিখুঁত হবে এবং ভবিষ্যদ্বাণী বা প্রেডিকশন তত বেশি বাস্তবসম্মত ও একুরেট হবে।

**৪. Large error হলে Cost Function এর value কেমন হবে?**

Large error হলে Cost Function-এর মান অনেক বেশি (High) হবে। এর অর্থ হলো মডেলের প্রেডিকশন খুব খারাপ হচ্ছে।

**৫. Cost কমানোর process এর নাম কী?**

Cost function-এর মান কমানোর বা মিনিমাইজ করার এই প্রসেসকে বলা হয় **Optimization Algorithm**, যার সবচেয়ে জনপ্রিয় উদাহরণ হলো **Gradient Descent**।

```python
# sklearn ব্যবহার করে একটি simple linear regression model তৈরি করো

from sklearn.linear_model import LinearRegression

house_sizes = np.array([1200, 1500, 1800, 2100, 2500]).reshape(-1, 1)
house_prices = np.array([50, 65, 75, 90, 110])


model = LinearRegression()
model.fit(house_sizes, house_prices)

new_house_size = np.array([[2000]])
predicted_price = model.predict(new_house_size)
print(predicted_price)
```

```python
print(f"Slope/Coefficient : {model.coef_[0]:.4f}")
print(f"Intercept : {model.intercept_:.4f}")
print(f"Predicted Price: {predicted_price[0]:.2f}")
```
