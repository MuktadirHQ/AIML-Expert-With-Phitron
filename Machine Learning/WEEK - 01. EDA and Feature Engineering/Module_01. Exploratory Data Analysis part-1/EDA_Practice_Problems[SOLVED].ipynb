Task 7:
Let's visualize the gender distribution among the surveyed penguins.
*   Extract the value counts for the sex column.
*   Create a pie chart using Matplotlib showing the male vs. female distribution.
*   Ensure the pie chart includes percentage labels with one decimal point (example 50.5%). Add a title to your chart.

```python
count = df['sex'].value_counts()
print(count)
```

```python
label = df['sex'].unique()[:-1]

plt.pie(count, labels=label, autopct="%1.1f")
plt.title("Gender distribution among the surveyed penguins")
plt.show()
```
