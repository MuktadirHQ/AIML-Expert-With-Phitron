**1. Countplot এবং Barplot এর মধ্যে পার্থক্য কী? কখন কোনটা use করবে?**

১. Countplot (কাউন্টপ্লট)
Countplot মূলত কোনো নির্দিষ্ট ক্যাটাগরির ডেটা কতবার ঘটেছে (frequency বা count), তা গণনা করে চার্টে দেখায়। এটি অনেকটা হিস্টোগ্রামের (Histogram) মতোই কাজ করে, তবে কন্টিনিউয়াস ডেটার বদলে ক্যাটেগরিক্যাল ডেটার জন্য ব্যবহৃত হয়।

Y-অক্ষ (Y-axis): সব সময় ওই ক্যাটাগরির মোট সংখ্যা (Count) নির্দেশ করে। এই মানটি ফাংশন নিজেই ডেটা থেকে গুনে বের করে নেয়।

কখন ব্যবহার করবেন: যখন আপনার কাছে শুধুমাত্র একটি ক্যাটেগরিক্যাল কলাম থাকবে এবং আপনি দেখতে চান কোন ক্যাটাগরির ডেটা কতগুলো আছে (How many?)।

ব্যবহারিক উদাহরণ: ধরুন, আপনি একটি ওয়েব স্ক্র্যাপিং স্ক্রিপ্ট রান করেছেন। এখন আপনি দেখতে চান আপনার রিকোয়েস্টগুলোর মধ্যে কতগুলো 200 OK, কতগুলো 404 Not Found এবং কতগুলো 500 Server Error এসেছে। এখানে শুধু স্ট্যাটাস কোডগুলোর মোট সংখ্যা গুনতে countplot ব্যবহার করা সবচেয়ে সহজ।

২. Barplot (বারপ্লট)
Barplot একটি ক্যাটেগরিক্যাল ভ্যারিয়েবলের বিপরীতে একটি নিউমেরিক্যাল (Numerical) ভ্যারিয়েবলের সম্পর্ক দেখায়। ডিফল্টভাবে, এটি প্রতিটি ক্যাটাগরির জন্য নিউমেরিক্যাল ডেটার গড় (Mean) বের করে বার হিসেবে প্রদর্শন করে (যদিও আপনি চাইলে mean-এর বদলে sum, median ইত্যাদিও সেট করতে পারেন)।

Y-অক্ষ (Y-axis): কোনো নিউমেরিক্যাল ভ্যালু (যেমন: গড়, যোগফল বা মিডিয়ান) নির্দেশ করে, যা ডেটাসেটে আগে থেকেই থাকতে হয়।

কখন ব্যবহার করবেন: যখন আপনি বিভিন্ন ক্যাটাগরির মধ্যে কোনো নিউমেরিক্যাল মানের তুলনা করতে চান (How much on average?)।

ব্যবহারিক উদাহরণ: ধরুন, আপনি একটি ট্রেডিং স্ক্রিপ্ট তৈরি করেছেন যা বিভিন্ন এক্সচেঞ্জ থেকে ডেটা নেয়। এখন আপনি Binance এবং ByBit-এর মতো এক্সচেঞ্জগুলোর "Average Daily Trading Volume" বা এপিআই-এর "Average Response Time" তুলনা করতে চান। এখানে ক্যাটাগরি হলো এক্সচেঞ্জের নাম এবং নিউমেরিক্যাল ডেটা হলো ভলিউম বা রেসপন্স টাইম। এই গড় মানগুলো তুলনা করতে barplot ব্যবহার করতে হবে।

**3. Y Data Profiling কেন manually EDA করার চেয়ে সুবিধাজনক? কী কী information এক জায়গায় পাওয়া যায়?**

Python-এ ডেটা নিয়ে কাজ করার সময় ম্যানুয়াল EDA (Exploratory Data Analysis) করাটা বেশ সময়সাপেক্ষ। যখন বড় কোনো ওয়েব স্ক্র্যাপিং প্রজেক্ট থেকে ডেটা এক্সট্র্যাক্ট করা হয় বা ট্রেডিংয়ের জন্য হিস্টোরিক্যাল ফাইন্যান্সিয়াল ডেটা প্রসেস করা হয়, তখন প্রতিটি কলামের ডিস্ট্রিবিউশন দেখতে বা মিসিং ভ্যালু খুঁজতে বারবার df.describe(), df.info(), বা Matplotlib/Seaborn-এর কোড লেখাটা বেশ ক্লান্তিকর এবং এতে ডেভেলপমেন্টের অনেকটা সময় নষ্ট হয়।

YData Profiling (আগে যার নাম ছিল Pandas Profiling) মূলত এই পুরো প্রক্রিয়াটিকে অটোমেট করে দেয়। মাত্র এক বা দুই লাইন কোড লিখেই পুরো ডেটাসেটের একটি বিস্তারিত, ইন্টারঅ্যাকটিভ HTML রিপোর্ট তৈরি করা যায়।

ম্যানুয়াল EDA-এর চেয়ে এটি কেন বেশি সুবিধাজনক এবং এখান থেকে কী কী তথ্য পাওয়া যায়, তা নিচে তুলে ধরা হলো:

কেন এটি ম্যানুয়াল EDA-এর চেয়ে সুবিধাজনক?
সময়ের সাশ্রয়: ম্যানুয়ালি ১০০ লাইনের কোড লিখে যে ইনসাইটগুলো বের করতে কয়েক ঘণ্টা সময় লাগতে পারে, YData Profiling মাত্র একটি কমান্ড (ProfileReport(df)) ব্যবহার করে কয়েক সেকেন্ড বা মিনিটের মধ্যে সেই একই কাজ করে দেয়।

স্ট্যান্ডার্ডাইজেশন এবং কমপ্লিটনেস: ম্যানুয়ালি কাজ করার সময় আমরা অনেক সময় কিছু ভ্যারিয়েবলের চেক করতে ভুলে যেতে পারি। কিন্তু এই টুলটি ডেটাসেটের প্রতিটি কলাম ধরে ধরে সিস্টেমেটিকভাবে বিশ্লেষণ করে, তাই কোনো কিছুই বাদ পড়ে না।

সহজে শেয়ার করার সুবিধা: এটি একটি সুন্দর ইন্টারঅ্যাকটিভ HTML রিপোর্ট জেনারেট করে। ক্লায়েন্ট বা টিমের অন্য সদস্যদের কাছে ডেটাসেটের অবস্থা প্রেজেন্ট করার জন্য এটি দারুণ একটি উপায়, কারণ রিপোর্টটি পড়ার জন্য তাদের কোড রান করার প্রয়োজন হয় না।

ইমিডিয়েট ওয়ার্নিং সিস্টেম: ডেটাসেটে কোনো বড় ধরনের সমস্যা থাকলে এটি একদম শুরুতেই অ্যালার্ট দিয়ে দেয়, যা ম্যানুয়াল পদ্ধতিতে অনেক সময় পরে গিয়ে ধরা পড়ে।

YData Profiling রিপোর্টে এক জায়গায় কী কী Information পাওয়া যায়?
রিপোর্টটি মূলত কয়েকটি সেকশনে ভাগ করা থাকে:

১. Overview (সারসংক্ষেপ)

ডেটাসেটে মোট কতগুলো ভ্যারিয়েবল (কলাম) এবং কতগুলো অবজারভেশন (সারি) আছে।

ডেটাসেটে মোট মিসিং সেলের (Missing cells) সংখ্যা এবং শতকরা হার।

ডুপ্লিকেট সারি (Duplicate rows) আছে কিনা।

ডেটাসেটটি মেমোরিতে কতটুকু জায়গা (Memory footprint) নিচ্ছে।

২. Alerts / Warnings (সতর্কবার্তা)
এটি সবচেয়ে গুরুত্বপূর্ণ অংশ। এখানে ডেটার বিভিন্ন সমস্যা হাইলাইট করা হয়:

High Cardinality: কোনো ক্যাটেগরিক্যাল কলামে অনেক বেশি ইউনিক ভ্যালু থাকলে (যেমন: ইউজারের নাম বা আইডি)।

High Correlation: দুটি কলামের মধ্যে খুব স্ট্রং সম্পর্ক থাকলে (যেমন: অ্যালগরিদমিক ট্রেডিংয়ের ক্ষেত্রে মাল্টি-কোলিনিয়ারিটি এড়াতে এটি জানা খুব জরুরি)।

Missing Values: কোনো কলামে প্রচুর ডেটা মিসিং থাকলে।

Zeroes: কোনো কলামে শূন্যের (0) পরিমাণ অস্বাভাবিক বেশি থাকলে।

৩. Variables (প্রতিটি কলামের বিস্তারিত অ্যানালিসিস)
প্রতিটি কলামের ডেটা টাইপ অনুযায়ী আলাদা বিশ্লেষণ:

নিউমেরিক্যাল ডেটার জন্য: Mean, Median, Min, Max, Standard Deviation, এবং ডেটার ডিস্ট্রিবিউশন দেখানোর জন্য স্বয়ংক্রিয় হিস্টোগ্রাম (Histogram)।

ক্যাটেগরিক্যাল ডেটার জন্য: ইউনিক ভ্যালুর সংখ্যা, সবচেয়ে বেশিবার আসা ভ্যালুগুলো (Mode), এবং তার কাউন্টপ্লট।

৪. Interactions (ইন্টারঅ্যাকশন)

দুটি নিউমেরিক্যাল ভ্যারিয়েবলের মধ্যে সম্পর্ক কেমন, তা দেখার জন্য অটোমেটেড স্ক্যাটারপ্লট (Scatterplot)। আপনি চাইলেই ড্রপডাউন থেকে ভ্যারিয়েবল পরিবর্তন করে সম্পর্ক দেখতে পারেন।

৫. Correlations (কোরিলেশন হিটম্যাপ)

Pearson, Spearman, Kendall-এর মতো বিভিন্ন স্ট্যাটিস্টিক্যাল মেথড ব্যবহার করে পুরো ডেটাসেটের একটি কোরিলেশন হিটম্যাপ তৈরি করে দেয়। এর মাধ্যমে খুব সহজেই বোঝা যায় কোন ভ্যারিয়েবলটি অন্যটির উপর কতটা নির্ভরশীল।

৬. Missing Values (মিসিং ভ্যালু অ্যানালিসিস)

বার চার্ট (Bar chart) এবং ম্যাট্রিক্সের (Matrix) মাধ্যমে ডেটাসেটের ঠিক কোথায় কোথায় ভ্যালু মিসিং আছে, তার একটি ক্লিয়ার ভিজ্যুয়াল রিপ্রেজেন্টেশন পাওয়া যায়।

মাত্র এক লাইনের কোড দিয়ে এতগুলো ইনসাইট পাওয়া যায় বলেই প্রিলিমিনারি ডেটা অ্যানালিসিসের ক্ষেত্রে এটি ডেভেলপার এবং ডেটা সায়েন্টিস্টদের প্রথম পছন্দ।

**4. নিচের code টি একটি Countplot তৈরি করবে যেখানে passenger class অনুযায়ী count দেখাবে এবং Gender দিয়ে আলাদা করবে।**
