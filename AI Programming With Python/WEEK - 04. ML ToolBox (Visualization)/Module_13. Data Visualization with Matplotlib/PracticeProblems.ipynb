# TASK 1 : Add New Column and Save File
*   You have a DataFrame with column Price.
*   Create a new column Discounted_Price with (10% discounted price).
*   Save the modified DataFrame as discounted_products.csv.

```python
df = pd.read_csv("/content/Practice.csv")
df
```

```python
df['Discounted_Price'] = df['Price'] - (df['Price'] * 0.10)
df
```

```python
df.to_csv("discounted_products.csv")
```

# TASK 2 : Check Duplicates and Remove Them

    Column: EmployeeID
Tasks:
*   Count duplicate entries
*   Remove duplicates
*   Display cleaned DataFrame

```python
dup_cnt = df.duplicated().sum()
print(dup_cnt)

df.drop_duplicates(subset='EmployeeID', inplace=True, keep='first')
df
```

# TASK 3 : Handle Missing Values
Columns: Age, Salary

Tasks:
*   Count missing values
*   Fill missing Age with mean age
*   Drop rows where Salary is missing

```python
missing_cnt = df[['Age', 'Salary']].isnull().sum()
print(missing_cnt)

df['Age'] = df['Age'].fillna(df['Age'].mean())

df.dropna(subset='Salary', inplace=True)
df
```

# TASK 4 : Apply Statistical Functions
Columns: Marks1, Marks2, Marks3

Tasks:
*   Calculate mean, max for each subject
*   Compute total marks per student
*   Find standard deviation of total marks

```python
#1
subjects = ["Marks1", "Marks2", "Marks3"]
stats = df[subjects].agg(['mean', 'max'])
print(stats)

#2
df['Total_Marks'] = df[subjects].sum(1)
display(df)

#3
print(df['Total_Marks'].std())


```
