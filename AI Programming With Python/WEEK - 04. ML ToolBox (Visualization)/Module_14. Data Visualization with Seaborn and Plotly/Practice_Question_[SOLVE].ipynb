```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sales = pd.read_csv("/content/sales.csv")
sales.head()
```

```python
students = pd.read_csv("/content/students.csv")
students.head()
```

##Problem : 1

```python
sns.lineplot(
    data=sales,
    x='date',
    y='sales'
)

```

##Problem : 2

```python
sns.scatterplot(
    data=students,
    x="hours_studied",
    y='marks',
    hue='gender'
)
```

##Problem : 3

```python
sns.relplot(
    data=students,
    kind='scatter',
    x='hours_studied',
    y='marks',
    col="subject"
)
```

##Problem : 4

```python
sns.histplot(
    data=students,
    kde=True,
    x='marks'
)
```

##Problem : 5

```python
sns.kdeplot(
    data=students,
    x='hours_studied',
    fill=True
)
```

##Problem : 6

```python
sns.countplot(
    data=students,
    x='subject'
)
```

##Problem : 7

```python
sns.regplot(
    data=students,
    x='hours_studied',
    y='marks'
)
```

##Problem : 8

```python
sns.pairplot(
    data=students,
    kind='scatter',
    hue='gender'
)
```
