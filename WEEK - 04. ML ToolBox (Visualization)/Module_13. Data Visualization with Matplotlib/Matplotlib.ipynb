**Customizing Line plot ( color,linestyle,marker) and Grid**

```python
plt.xlabel("Year")
plt.ylabel("Count of Enrollment")
plt.title("Student Enrollment over 8 years")

plt.plot(df['Year'], df['Programming'],
         label='Programming',
         color="red",
         linewidth=5,
         linestyle='dashdot',
         marker='o',
         markersize=10)

plt.plot(df['Year'], df['Digital Marketing'],
         label='Digital Marketing',
         color='#08f4fc',
         linewidth=3,
         linestyle='dotted',
         marker='x',
         markersize=10)



plt.legend()
plt.show()
```

```python
plt.xlabel("Year")
plt.ylabel("Count of Enrollment")
plt.title("Student Enrollment over 8 years")

plt.plot(df['Year'], df['Programming'],
         label='Programming',
         color="red",
         linewidth=5,
         linestyle='dashdot',
         marker='o',
         markersize=10)

plt.plot(df['Year'], df['Digital Marketing'],
         label='Digital Marketing',
         color='black',
         linewidth=3,
         linestyle='dotted',
         marker='x',
         markersize=10)



plt.legend()
plt.grid()
plt.show()
```

**Usage of Scatter Plots**

#Scatter Plots

```python
students = pd.read_csv("/content/student_IQdata.csv")
students
```

```python
# Relation between study hours and IQ score --> Positive Correlation.

x = students['Study_Hour']
y = students['IQ_Score']

plt.xlabel("Study Hours")
plt.ylabel("IQ Score")
plt.title("Relation Between Study Hours and IQ Score")
plt.scatter(x, y)
```

```python
# Relation between Shoe_Size and IQ score --> NO Correlation.

x = students['Shoe_Size']
y = students['IQ_Score']

plt.xlabel("Shoe Size")
plt.ylabel("IQ Score")
plt.title("Relation Between Shoe Size and IQ Score")
plt.scatter(x, y)
```

```python
# Relation between Chilling Hours and IQ score --> Negative Correlation.

x = students['Chilling_Hours']
y = students['IQ_Score']

plt.xlabel("Chilling Hours")
plt.ylabel("IQ Score")
plt.title("Relation Between Chilling Hours and IQ Score")
plt.scatter(x, y)
```

#Histogram
*   data distributed over a factor.
*   frequency.
*   outliers.

```python
#Based on IQ

x = students['IQ_Score']
plt.hist(x, edgecolor='black')
plt.xlabel("IQ Score")
plt.ylabel("Frequency")
```

```python
#Based on Study Hours

x = students['Study_Hour']
plt.hist(x, edgecolor='black', bins=[1, 2, 3, 4, 5]) #bins=15
plt.xlabel("Study Hours")
plt.ylabel("Frequency")
```

#Bar Chart

*   Categorical

```python
data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
df
```

```python
# Gender wise

gender_group = df.groupby("Gender").size()
gender_group.index
gender_group.values
```

```python
x = gender_group.index
y = gender_group.values

plt.xlabel("Gender")
plt.ylabel("Count")
plt.title("Barchart of Gender")
plt.bar(x, y, color=['orange', 'blue'], edgecolor='black')
```

```python

df =pd.read_csv("/content/student_data.csv")
df
```

```python
# Bar chart on Real Data

status = df.groupby("CompletionStatus").size()

x = status.index
y = status.values

plt.xlabel("Completion Status")
plt.ylabel("Count")
plt.title("Barchart of Course Completion")
plt.bar(x, y, color=['green', 'yellow', 'red'], edgecolor='black')
plt.show()
```

```python
# Location

status = df.groupby("Location").size()

x = status.index
y = status.values

plt.xlabel("Locatoin")
plt.ylabel("Count")
plt.title("Barchart of Location Student")
plt.bar(x, y, edgecolor='black')
plt.show()
```

# Pie Chart
*   Categorical / Neumerical
*   Percentage

```python
status = df.groupby("CompletionStatus").size()

plt.pie(status, labels=status.index, autopct='%1.1f%%', explode=(0.05, 0, 0), shadow=True)
plt.show()
```
